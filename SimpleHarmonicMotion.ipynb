```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
'''

WHat is Simple Harmonic Motion (SHM)?
Simple Harmonic Motion (SHM) is a type of periodic motion where an object oscillates
back and forth around an equilibrium position.
In SHM The restoring force is directy proportional to the negative displacement of the particle from the equilibrium position.

In this code, we will draw the graphs of displacement, velocity and acceleration of the SHM of a spring mass system.
Along with the potential and kinetic energy curve of the particle in SHM.
Also Later we will draw the phase space diagram of the particle in SHM.

'''
```

```python
#The differential equation of SHM is given by:
# m * d²x/dt² + k * x = 0
```

```python
#DEFINING VARIABLES
m = 0.5 # mass of particle in kg
k = 10 # spring constant in N/m
omega = np.sqrt(k/m) #angular frequency in rad/s
T = 2 * np.pi / omega #period of oscillation in s
```

```python
#POSITION - Time graph of particle in SHM
x = 0.1 * np.cos(omega * np.linspace(0, 2*T, 100)) #displacement of particle in m
t = np.linspace(0, 2*T, 100) #time in seconds
plt.figure(figsize=(10, 5))
plt.grid(True, color = 'black', linestyle = '--', linewidth = 0.8)
plt.xlabel('Time (s)')
plt.ylabel('Displacement (m)')
plt.title('Displacement-Time graph')
plt.plot(t,x, color = 'red')    

```

```python
#Velocity - Time graph of a particle in SHM
y = 0.1 * omega * np.sin(omega* np.linspace(0, 2*T, 100)) #velocity of particle in m/s
t = np.linspace(0, 2*T, 100) #time in seconds
plt.grid(True, color ='black', linestyle = '--', linewidth = 0.8)
plt.xlabel("Time(s)")
plt.ylabel("Velocity(m/s)")
plt.title('Velocity Time Graph')
plt.plot(t,y,color = 'green')
```

```python
#Velocity-Position graph of the particle
y = 0.1 * omega * np.sin(omega* np.linspace(0, 2*T, 100)) #velocity of particle in m/s
x = 0.1 * np.cos(omega * np.linspace(0, 2*T, 100)) #displacement of particle in m
plt.grid(True, color ='black', linestyle = '--', linewidth = 0.8)
plt.xlabel("Position(m)")
plt.ylabel("Velocity(m/s)")
plt.title('Velocity Position Graph')
plt.plot(x,y,color = 'black')
```
