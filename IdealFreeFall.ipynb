```python
#IMPORTING MODULES
import numpy as np
import matplotlib.pyplot as plt
import math
```

```python
#DEFINING VARIABLES AND CONSTANT g = 9.81 m/s^2
t = np.linspace(0, 10, 100)
v_0 = 0
g = 9.81
x_0 = v_0 * t + 0.5 * g * t**2
```

```python
#POSITION TIME GRAPH OF A PARTICLE UNDER FREE FALL
plt.plot(t, x_0, color = 'red')
plt.grid(True, color = 'black', linestyle = '--', linewidth = 0.8)
plt.title('Ideal Free Fall')
plt.xlabel('Time(s)')
plt.ylabel('Position(m)')
plt.show()
```

```python
#VELOCITY TIME GRAPH OF A PARTICLE UNDER FREE FALL
v = v_0 + g*t
plt.plot(t, v, color='red')
plt.grid(True, color = 'black', linestyle = '--', linewidth = 0.8)
plt.title('Velocity vs Time')
plt.xlabel('Time(s)')
plt.ylabel('Velocity(m/s)')
plt.show()
```

```python
#POSITION-VELOCITY GRAPH OF A PARTICLE UNDER FREE FALL
plt.plot(x_0, v, color = 'red')
plt.grid(True, color = 'black', linestyle = '--', linewidth = 0.8)
plt.title('Position vs Velocity')
plt.xlabel('Position(m)')
plt.ylabel('Velocity(m/s)')
plt.show()
```

```python
#DEFINING EULER'S METHOD FOR FREE FALL

def euler_free_fall(x0, v0, g, dt, t_max):
    t = np.arange(0, t_max, dt)
    x = np.zeros(len(t))
    v = np.zeros(len(t))
    x[0] = x0
    v[0] = v0
    
    for i in range(1, len(t)):
        v[i] = v[i-1] + g * dt
        x[i] = x[i-1] + v[i-1] * dt
        
    return t, x, v  
```

```python
#USING EULER'S METHOD TO SIMULATE FREE FALL
dt = 0.1
x0 = 0
v0 = 0
t_max = 10
t_euler, x_euler, v_euler = euler_free_fall(x0, v0, g, dt, t_max)

#PLOTTING RESULTS FROM EULER'S METHOD
plt.plot(t_euler, x_euler, label='Position (Euler)', color='red')
plt.plot(t_euler, v_euler, label='Velocity (Euler)', color='blue')
plt.title('Euler Method Simulation of Free Fall')
plt.xlabel('Time(s)')
plt.ylabel('Position(m) / Velocity(m/s)')
plt.legend()
plt.grid(True, color = 'black', linestyle = '--', linewidth = 0.8)
plt.show()
```

```python
#LETS NOW PLOT THE RESULTS FROM EULER'S METHOD ALONG WITH THE ANALYTICAL SOLUTION FOR COMPARISON
plt.plot(t, x_0, label='Position (Analytical)', color = 'red', linestyle='-')
plt.plot(t_euler, x_euler, label='Position (Euler)', color='blue')
plt.title('Position vs Time: Analytical vs Euler Method')
plt.grid(True, color = 'black', linestyle = '--', linewidth = 0.8)
plt.show()
```

```python
#WE OBSERVE THAT THE EULER METHOD APPROXIMATION IS CLOSE TO THE ANALYTICAL SOLUTION,
#BUT THERE IS A SMALL ERROR DUE TO THE NUMERICAL APPROXIMATION.
```
